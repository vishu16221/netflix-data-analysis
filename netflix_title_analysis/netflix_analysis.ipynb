```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV
df = pd.read_csv('E:/netflix_titles_cleaned.csv')

# Check the data
print(df.info())
print(df.head())
```

```python
# Movies vs TV Shows
type_distribution = df['type'].value_counts()
print(type_distribution)

# Top genres (splitting the comma-separated list)
genres = df['listed_in'].str.split(', ', expand=True).stack()
top_genres = genres.value_counts().head(10)
print(top_genres)
```

```python
# Bar chart: Movies vs TV Shows
plt.figure(figsize=(8, 5))
type_distribution.plot(kind='bar', color=['skyblue', 'coral'])
plt.title('Netflix Content Distribution: Movies vs TV Shows')
plt.xlabel('Type')
plt.ylabel('Number of Titles')
plt.xticks(rotation=0)
plt.show()
```

```python
# Bar chart: Top genres
plt.figure(figsize=(12, 6))
top_genres.plot(kind='bar', color='teal')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

```python
# Export cleaned data for Power BI
df.to_csv('E:/practice/netflix_title_analysis/netflix_for_powerbi.csv', index=False)
print("Data exported for Power BI!")
```
